# 1) Area of a Rectangle:

### Write a Python program to calculate the area of a rectangle. Use the formula:
### 𝐴𝒓𝒆𝒂 = 𝒍𝒆𝒏𝒈𝒕𝒉 × 𝒘𝒊𝒅𝒕𝒉

# 7) Compound Interest:

### Write a Python program to calculate compound interest. Use the formula:
### A = P * (1 + r/n)^n-t
### Where:
### A= total amount
### P = principal amount
### r = annual interest rate (decimal)
### n = number of times interest is compounded per year
### t = time in years
### Take P, r, n, and t as inputs from the user.

```python
p = float(input("Enter the principal amount:"))
r = float(input("Enter the annual interest rate (decimal):"))
n = float(input("Enter the  number of times interest is compounded per year:"))
t = float(input("Enter the time in years:"))
total_amount = p*(1+r/n)**n-t
print("Total amount is" ,total_amount)
```

# 11) Quadratic Equation Roots:

### Write a Python program to find the roots of a quadratic equation. Use the formula:
### X = -b±√b^2 - 4ac / 2a
### Take a, b, and c as inputs from the user.

```python
import math
# First we find (+) value of Quadratic Equation:
a = float(input("Enter the value of a:"))
b = float(input("Enter the value of b:"))
c = float(input("Enter the value of c:"))
d = (b**2) - 4*a*c
x1 = (-b + math.sqrt(d))/(2*a)
print("The (+) value of equation is:", x1)
# Now we find (-) value of Quaratic Equation:
x = float(input("Enter the value of x:"))
y = float(input("Enter the value of y:"))
z = float(input("Enter the value of z:"))
e = (y**2) - 4*x*z
x2 = (-y - math.sqrt(e))/(2*x)
print("The (-) value of equation is:", x2)
# Note: Error may cause, when the value of square root is Negative(-).
```

# 19) Work Done by a Force:

### Write a Python program to calculate the work done by a force. Use the formula:
### W = f.d Cos(theta)
### where:
### W = work done
### f= force
### d = distance
### theta = angle between force and direction of movement (in degrees)
### Take f, d, and θ as inputs from the user.

```python
import math
force = float(input("Enter the Force:"))
distance = float(input("Enter the Distance:"))
theta = float(input("Enter the angle between force and direction of movement (in degrees):"))
work = (force*distance) * math.cos(theta)
print("The work done by a force is:", work)
```

# 20) Heat Transfer:

### Write a Python program to calculate the amount of heat transferred. Use the formula:
### 𝑄 = 𝑚𝑐 (ΔT)
### where:
### Q= heat transfer
### m = mass
### c = specific heat capacity
### ΔT
### Take m, c, T as inputs from the user.

```python
m = float(input("Enter the mass:"))
c = float(input("Enter the  specific heat capacity:"))
T_initial = float(input("Enter the value of initial temperature:"))
T_final = float(input("Enter the value of final temperature:"))
dt = T_final - T_initial
Q = (m*c) *dt
print("The amount of heat transferred is: ", Q)
```
