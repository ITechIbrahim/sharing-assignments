## A student is buying a laptop.

    Laptop price: Rs. 85,000
    Student has: Rs. 100,000
    Student gets a discount of Rs. 5,000
    A laptop bag costs Rs. 2,500
    The store has a promotion code "SAVE10".

### Create variables and:

    Calculate the laptop price after discount.
    Add the bag price to get the final bill.
    Calculate the remaining money.
    Check whether the student has enough money.
    Check whether "SAVE10" is present in the promotion code.
    Check whether the final bill is greater than Rs. 80,000.
    Store the result of whether the student has enough money in a variable and check whether it is truthy.

```python
laptop_price = 85000
student_has = 100000
discount = 5000
laptop_bag_cost = 2500
promotion_code = "SAVE10"
# Calculating the laptop price after discount.
discounted_laptop = laptop_price - discount
print("The laptop price after discount is",discounted_laptop)
# Adding the bag price to get the final bill.
final_bill = laptop_bag_cost + discounted_laptop
print("The bag price to get the final bill is", final_bill)
# Calculating the remaining money.
remaining_money = student_has - final_bill
print("The remaining money is", remaining_money )
# Checking whether the student has enough money.
enough_money = student_has > final_bill
print("The student has enough money, thus its", enough_money)
# Checking whether "SAVE10" is present in the promotion code.
check_pro_code = promotion_code == "SAVE10"
print("'SAVE10' is present in the promotion code, therefore its", check_pro_code)
# Checking whether the final bill is greater than Rs. 80,000.
bill = final_bill > 80000
print("The final bill is greater than Rs. 80,000, thats why its", bill)
# Storing the result of whether the student has enough money in a variable and check whether it is truthy.
result = bool(student_has)
print("The result of whether the student has enough money in a variable and check whether it is truthy, therefore its", result)
```

### You are creating a student result calculator.

A student has:

    Mathematics: 85
    English: 72
    Python: 90
    Total possible marks per subject: 100
    Student name: "Ahmed"
    Required passing marks per subject: 50

### Create variables and:

    Calculate the student's total marks.
    Calculate the average marks.
    Calculate the percentage.
    Check whether the student passed Mathematics.
    Check whether the student passed all three subjects.
    Check whether the name contains the letter "A".
    Check whether the average marks are greater than or equal to 80.
    Store the final percentage and check whether it is truthy.

Print all results.

```python
# Calculating the student's total marks
name = "Ahmed"
math = 85
eng = 72
python = 90
passing_marks = 50
total_sub_marks = 100
obtained_marks = math + eng + python
print("The Total marks of student is",obtained_marks)
# Calculating the average marks.
avg_of_marks = obtained_marks / 3
print("The Average of marks is", avg_of_marks)
# Calculating the percentage.
percentage = obtained_marks / (total_sub_marks * 3) *100
print("The Precentage of student is",percentage)
# Checking whether the student passed Mathematics.
pass_in_math = math >= passing_marks
print("The student passed in Mathematics, thus its", pass_in_math)
# Checking whether the student passed all three subjects.
all_sub_pass = math >= 50 and python >= 50 and eng >= 50
print("The student passed all three subjects, thats why its", all_sub_pass)
# Checking whether the name contains the letter "A"
letter = "A" in name
print("The name contains the letter 'A', therefore its", letter)
# Checking whether the average marks are greater than or equal to 80
check_avg_marks = avg_of_marks == 80
print("The average marks aren't greater than or equal to 80, therefore its", check_avg_marks)
# Storing the final percentage and check whether it is truthy.
truthy_value = bool(percentage)
print("The final percentage and check whether it is truthy, therefore its", truthy_value)
```
