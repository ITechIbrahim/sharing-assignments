## Write a program to decide insurance eligibility.
Take input:
* age
* smoker (Yes/No)
* existing_disease (Yes/No)
* annual_income
#### Rules:
* age < 18 → reject
* smoker + disease → high premium category
* income < 200000 → limited coverage
* age > 60 → special medical check required
* otherwise → normal approval

```python
age = int(input("Enter your age: "))

if age < 18:
    print("Rejected")

else:
    smoker = input("Are you a smoker? (yes/no): ").lower()
    existing_disease = input("Existing disease? (yes/no): ").lower()
    annual_income = int(input("Enter your annual income: "))

    if smoker == "yes" and existing_disease == "yes":
        print("High premium category")

    elif annual_income < 200000:
        print("Limited coverage")

    elif age > 60:
        print("Special medical check required")

    else:
        print("Normal approval")
```

## Write a program to detect suspicious banking transactions.
Take input:
* transaction_amount
* location_match (Yes/No)
* time_of_transaction(Day/Night)
* previous_transactions_count
* device_verified (Yes/No)
#### Rules:
* amount > 100,000 → suspicious
* location mismatch + high amount → fraud alert
* night transaction + unverified device → high risk
* many transactions (>10 in 1 hour) → freeze account
* otherwise → normal transaction

```python
transaction_amount = int(input("Enter your transaction amount :"))
location = input("Location match (yes/no)").lower()
time_of_transaction = input("Enter your time of transaction (Day/Night)").lower()
previous_transaction_count = int(input("Enter your previous transaction count :"))
device_verified = input("Is this your device is verified (yes/no)").lower()
if transaction_amount > 100000:
    print("Suspicious")
elif location == 'no' and transaction_amount > 100000:
    print("Fraud Alert")
elif time_of_transaction == 'night' and device_verified == 'no':
    print("High Risk")
elif previous_transaction_count > 10 and '1hr':
    print("Freeze acount")
else:
    print("Normal Transaction")
```

## Write a program to classify a person’s risk level in a police database.
Take input:
* criminal_record (Yes/No)
* number_of_cases
* travel_history_flagged (Yes/No)
* financial_activity_score
* last_activity_days
#### Rules:
* criminal_record = Yes AND cases > 3 → High Risk
* flagged travel + suspicious financial score → High Risk
* no activity for long time (>365 days) → Monitor
* clean record + normal activity → Low Risk
* mixed signals → Medium Risk

```python
crinimal_record = input("Enter your crinimal record (yes/no):").lower()
no_of_cases = int(input("Enter your number of cases :"))
travel_history_flagged = input("Enter your travel history flagged (yes/no):").lower()
financial_activity_score = int(input("Enter your financial activity score :"))
last_activity_days = int(input("Enter your last activity days :"))
if crinimal_record == 'yes' and no_of_cases > 3:
    print("High risk")
elif travel_history_flagged == 'yes' and financial_activity_score >= 70:
    print("High Risk")
elif last_activity_days > 365:
    print("Monitor")
elif crinimal_record == 'no' and travel_history_flagged == 'no' and financial_activity_score < 70:
    print("Low risk")
else:
    print("Medium Risk")
```
