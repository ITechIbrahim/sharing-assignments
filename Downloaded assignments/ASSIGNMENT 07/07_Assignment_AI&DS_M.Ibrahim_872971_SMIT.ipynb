## 🅿️ 10. Parking Fee System

### Goal:
    Calculate the parking fee of a mall for different vehicles.

### Steps:

    Ask for the vehicle type (car or bike).
    
    Then ask for the number of hours parked.
    
    Car: Rs 100 for up to 2 hours, plus Rs 50 for each extra hour.
    
    Bike: Rs 50 for up to 2 hours, plus Rs 20 for each extra hour.
    
    Any other vehicle → "Vehicle not allowed".

### Example:

    Enter vehicle: car
    Enter hours: 5
    Parking fee = Rs 250

```python
vehicle_type = input("Enter your vehicle type (car or bike)").lower()
if vehicle_type not in ['bike', 'car']:
    print("Vehicle not allowed")
else:
    no_of_hr = int(input("Enter number of hours parked :"))
    if vehicle_type == 'car':
        if no_of_hr <= 2:
            parking_fee = 100
        else:
            parking_fee = 100 + (no_of_hr - 2) * 50
    elif vehicle_type == 'bike':
            if no_of_hr <= 2:
                parking_fee = 50
            else:
                parking_fee = 50 + (no_of_hr - 2) * 20
print(f"The vehicle type is {vehicle_type}, number of hours {no_of_hr} and parking fees is {parking_fee}")
```

### Steps:

    Ask for the ride type (economy or luxury).
    
    Ask for the distance in km.
    
    Economy: Rs 200 for up to 5 km, then Rs 30 per extra km.
    
    Luxury: Rs 400 for up to 5 km, then Rs 60 per extra km.
    
    Print the final fare.

### Example:

    Enter ride type: economy
    Enter distance: 8
    Fare = Rs 290

```python
ride_type = input("Enter your ride type (economy or luxury):").lower()
if ride_type not in ['economy', 'luxury']:
    print("Ride type is not found ")
else:
    distance = float(input("Enter the distance in km :"))
    if ride_type == 'economy':
        if distance <= 5:
            final_fee = 200
        else:
            final_fee = 200 + (distance - 5) * 30
    elif ride_type == 'luxury':
        if distance <= 5:
            final_fee = 400
        else:
            final_fee = 400 + (distance - 5) * 60
print(f"Your ride type is '{ride_type}' and your final fare is {final_fee}PKR and distance {distance}km")
```

    The answer is the final value of a (the GCD).

### Example:

## 🌡️ 50. Weekly Temperature Report

### Goal:
    A weather station wants to know how many days were warmer than average.

### Steps:

    Create temps = [30, 32, 28, 35, 31, 29, 36].
    
    First loop: add all the temperatures and calculate the average.
    
    Second loop: count the days above the average.
    
    Print the average (2 decimals) and the count.

### Example:

    Average = 31.57
    Days above average = 3

```python
create_temps = [30, 32, 28, 35, 31, 29, 36]
total = 0
for t in create_temps:
    total += t
avg = total / len(create_temps)
count = 0
for t in create_temps:
    if t > avg:
        count += 1
    
print(f"Average = {round(avg, 2)}")
print(f"Days above average = {count}")
```
